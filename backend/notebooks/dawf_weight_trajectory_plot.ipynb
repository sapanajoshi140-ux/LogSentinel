```python
import pandas as pd
import matplotlib.pyplot as plt

# Load DAWF weight trajectory
df = pd.read_csv("../results/hdfs/dawf_weight_trajectory.csv")

# Plot weights
plt.figure(figsize=(10, 5))

plt.plot(
    df["step"],
    df["ewma_weight"],
    label="EWMA Weight"
)

plt.plot(
    df["step"],
    df["markov_weight"],
    label="Markov Weight"
)

plt.xlabel("Test Sequence")
plt.ylabel("DAWF Weight")
plt.title("DAWF Weight Trajectory on HDFS Test Split")

plt.legend()
plt.grid(True)
plt.tight_layout()

plt.show()
```
